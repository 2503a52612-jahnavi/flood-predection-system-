```python
import pandas as pd
```

```python
df = pd.read_csv('D:/DAV.project/CLEAN_FLOOD_DATA_CLEANED.csv')
df.head()
```

```python
df = df.drop_duplicates()
print('Rows after removing duplicates:', len(df))
```

```python
lives_lost = df.groupby("Name of State/UT")["Human Lives Lost Nos."].sum().reset_index()
lives_lost = lives_lost.sort_values("Human Lives Lost Nos.", ascending=False)

# Keep top 10, group rest
top10 = lives_lost.head(10)
others = pd.DataFrame({
    "Name of State/UT": ["Others"],
    "Human Lives Lost Nos.": [lives_lost["Human Lives Lost Nos."].iloc[10:].sum()]
})

lives_lost_top = pd.concat([top10, others])

fig = px.pie(
    lives_lost_top,
    names="Name of State/UT",
    values="Human Lives Lost Nos.",
    title="Top 10 States by Human Lives Lost (2016–2018)",
    hole=0.2
)
fig.show()

```

```python
fig = px.histogram(
    df,
    x="Area Affected (Mha)",
    nbins=30,
    color="Year",
    title="Distribution of Area Affected (Mha)"
)
fig.update_traces(opacity=0.7)
fig.show()

```

```python
pop_year = df.groupby("Year")["Population Affected (Million)"].sum().reset_index()

fig = px.line(
    pop_year,
    x="Year",
    y="Population Affected (Million)",
    markers=True,
    title="Yearly Trend of Population Affected"
)
fig.show()
```

```python
crop_damage = df.groupby("Name of State/UT")["Damages to Crops - Value (Rs. Crore)"].sum().reset_index()
crop_damage = crop_damage.sort_values("Damages to Crops - Value (Rs. Crore)", ascending=False).head(10)

fig = px.bar(
    crop_damage,
    x="Name of State/UT",
    y="Damages to Crops - Value (Rs. Crore)",
    text_auto=True,
    title="Top 10 States by Crop Damage"
)
fig.show()
```

```python
fig = px.scatter(
    df,
    x="Damage to Houses - Value (Rs. Crore)",
    y="Population Affected (Million)",
    color="Year",
    hover_name="Name of State/UT",
    title="House Damage vs Population Affected"
)
fig.show()
```

```python
import pandas as pd
import plotly.express as px

df = pd.read_csv("CLEAN_FLOOD_DATA_CLEANED.csv")

# Rename columns for clarity
df_renamed = df.rename(columns={
    "Area Affected (Mha)": "Area Affected",
    "Population Affected (Million)": "Population Affected",
    "Damages to Crops - Value (Rs. Crore)": "Crop Damage Value",
    "Damage to Houses - Value (Rs. Crore)": "House Damage Value",
    "Human Lives Lost Nos.": "Lives Lost",
    "Total Damages Crops,Houses & Public Utilities (Rs. Crore)": "Total Damages"
})

# Correlation matrix
corr = df_renamed[
    ["Area Affected", "Population Affected", "Crop Damage Value",
     "House Damage Value", "Lives Lost", "Total Damages"]
].corr()

# Build smaller heatmap
fig = px.imshow(
    corr,
    text_auto=True,
    color_continuous_scale="RdBu_r",
    title="Correlation Heatmap of Flood Impact Variables",
    width=500,   # smaller width
    height=500   # smaller height
)

# Layout tweaks
fig.update_layout(
    font=dict(size=12),
    title_font=dict(size=18),
    xaxis=dict(side="top"),
    margin=dict(l=50, r=50, t=80, b=50)
)

fig.show()



```
