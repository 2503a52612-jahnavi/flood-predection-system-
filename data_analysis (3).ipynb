```python
import pandas as pd
```

```python
df = pd.read_csv('C:/Users/hp/OneDrive/Documents/Desktop/polution analytics/global air pollution dataset.csv')
df.head()
```

```python
# Remove duplicates
df = df.drop_duplicates()
print('Rows after removing duplicates:', len(df))
```

```python
import plotly.express as px

# 1. Prepare the data
aqi_category = df["AQI Category"].value_counts().reset_index()
aqi_category.columns = ["AQI Category", "Count"]

# 2. Build the interactive donut chart
fig = px.pie(
    aqi_category,
    names="AQI Category",
    values="Count",
    title="Distribution of AQI Categories",
    hole=0.3
)

# 3. Render the plot
fig.show()

```

```python
# 3. HISTOGRAM
# =======================================================

fig = px.histogram(
    df,
    x="AQI Value",
    title="AQI Distribution",
    nbins=30
)

fig.show()
```

```python
#scatter plot 
fig = px.scatter(
    df,
    x="PM2.5 AQI Value",
    y="AQI Value",
    color="AQI Category",
    hover_name="City",
    hover_data=["Country"],
    title="PM2.5 AQI vs Overall AQI"
)

fig.show()
```

```python
#Heat map

corr = df[
    [
        "AQI Value",
        "CO AQI Value",
        "Ozone AQI Value",
        "NO2 AQI Value",
        "PM2.5 AQI Value"
    ]
].corr()

fig = px.imshow(
    corr,
    text_auto=True,
    title="AQI and Pollutant Correlation"
)

fig.show()
```
